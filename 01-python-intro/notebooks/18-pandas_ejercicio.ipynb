```python
# 20. Eliminar valores faltantes 
# Crea un DataFrame llamado df_clean que elimine únicamente los registros que tengan valores faltantes en 
# sex o body_mass_g. Comprueba cuántas filas fueron eliminadas. 
df_clean=df.dropna(subset=["sex","body_mass_g"])
df_clean
```

```python
# 21. Crear una nueva columna 
# Crea una columna llamada body_mass_kg con la masa corporal expresada en kilogramos. Recuerda: 1000 
# g = 1 kg. No modifiques body_mass_g. 
df["body_mass_kg"]=df["body_mass_g"]/1000
df.head(10)
```

```python
# 22. Clasificación mediante pd.cut() 
# Crea una nueva columna mass_category clasificando los pingüinos como Light si pesan menos de 3500 g, 
# Medium si pesan entre 3500 y 4500 g (inclusive) y Heavy si pesan más de 4500 g. Utiliza pd.cut(). 
df["mass_category"] = pd.cut(
    df["body_mass_g"],
    bins=[-float("inf"), 3500, 4500, float("inf")],# rango de -infinito a 3500 a 4500 a mas inifinito
    labels=["Light", "Medium", "Heavy"]
)

df.head(10)
```

```python
# 23. Masa promedio por especie 
# Calcula la masa corporal promedio (body_mass_g) de cada especie. El resultado debe tener una fila por 
# especie. 
agrupado=df.groupby("species")["body_mass_g"].agg(Promedio ="mean")
agrupado
```

```python
# 24. Múltiples estadísticas 
# Para cada especie calcula sobre body_mass_g: cantidad, promedio, mínimo, máximo y desviación estándar. 
# Intenta resolverlo utilizando groupby() y agg(). 
agrupado=df.groupby("species")["body_mass_g"].agg([
    "count",
    "mean",
    "min",
    "max",
    "std"
])
agrupado
```

```python
# 25. Agrupación utilizando dos columnas 
# Calcula la masa corporal promedio agrupando por species y sex. El resultado debe permitir comparar 
# machos y hembras dentro de cada especie. 
agrupado=df.groupby(["species","sex"])["body_mass_g"].agg([
    "count",
    "mean",
    "min",
    "max",
    "std"
])
agrupado
```

```python
# 26. Tabla cruzada 
# Construye una tabla que muestre cuántos pingüinos existen para cada combinación species × island. 
# Puedes utilizar pd.crosstab() o pivot_table(). 
pd.crosstab(df["species"], df["island"])
```

```python
# 27. Pingüino más pesado de cada especie
# Encuentra el registro correspondiente al pingüino con mayor body_mass_g dentro de cada especie. El
# resultado final debe mostrar species, sex, island y body_mass_g. No basta con mostrar únicamente la masa
# máxima: debes recuperar el registro completo correspondiente al máximo.

indices=df.groupby("species")["body_mass_g"].idxmax()
df.loc[indices]
```

```python
# 28. Diferencia respecto al promedio 
# Crea una columna mass_difference que indique cuánto pesa cada pingüino por encima o por debajo del 
# promedio de su propia especie. Conceptualmente: masa del pingüino - masa promedio de su especie. 
# Intenta resolverlo usando groupby() y transform(). 
df["mass_difference"] = (
    df["body_mass_g"] -
    df.groupby("species")["body_mass_g"].transform("mean")#lo devuelve repetido en cada fila en cada elemento
)
df.head(10)
```

```python
# 29. Ranking dentro de cada especie 
# Crea una columna mass_rank que indique el ranking de masa corporal de cada pingüino dentro de su propia 
# especie. El pingüino más pesado debe tener mass_rank = 1. Luego muestra los 3 pingüinos más pesados de 
# cada especie. 
df["mass_rank"] = (
    df.groupby("species")["body_mass_g"].rank(method="min", ascending=False)
)

df[df["mass_rank"] <= 3].sort_values(["species", "mass_rank"])
```

```python
# 30. Ejercicio integrador — Reporte de especies
# Construye un DataFrame llamado species_report con una fila por especie y las columnas: species,
# total_penguins, average_body_mass, min_body_mass, max_body_mass, average_flipper_length y
# average_bill_length. Redondea los promedios a 2 decimales. Luego ordénalo de mayor a menor por
# average_body_mass, identifica la especie con mayor masa promedio y la especie con mayor longitud
# promedio de aleta, y guarda el resultado en species_report.csv sin incluir el índice.

# agrupado=df.groupby("species")["body_mass_g"].agg([
#     "count",
#     "mean",
#     "min",
#     "max",
#     "std"
# ])
dataframe=df.groupby("species").agg(
    total_pinguinos=("body_mass_g", "count"),
    average_body_mass=("body_mass_g", "mean"),
    min_body_mass=("body_mass_g", "min"),
    max_body_mass=("body_mass_g", "max"),
    average_flipper=("flipper_length_mm", "mean"),
    average_bill_length=("bill_length_mm", "mean")
).round(2).sort_values("average_body_mass",ascending=False)
dataframe
```

```python
species_report = df.groupby("species").agg(
    total_penguins=("body_mass_g", "count"),
    average_body_mass=("body_mass_g", "mean"),
    min_body_mass=("body_mass_g", "min"),
    max_body_mass=("body_mass_g", "max"),
    average_flipper_length=("flipper_length_mm", "mean"),
    average_bill_length=("bill_length_mm", "mean")
).round(2).sort_values("average_body_mass", ascending=False).reset_index()

species_report
```

```python
species_report.loc[species_report["average_body_mass"].idxmax(), "species"]
species_report.loc[species_report["average_flipper_length"].idxmax(), "species"]
```

```python
species_report.to_csv("species_report.csv", index=False)
```
