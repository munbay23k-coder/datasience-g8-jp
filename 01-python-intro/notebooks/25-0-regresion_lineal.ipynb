```python
import numpy as np
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt 

X = np.array([1, 2, 3, 4]).reshape(-1, 1)
y = np.array([2, 4, 5, 8])
 
plt.scatter(X,y)
```

```python
modelo = LinearRegression()
modelo.fit(X, y)# se entrena el modelo con x y 
 
```

```python

print(modelo.coef_)        # [1.9]  la pendiente
print(modelo.intercept_)   # 0.0   el punto de partida casi 0
```

```python
# cuando X es una lista simple reshape(1,1) la convierte en una columna
predictions=modelo.predict(X.reshape(-1,1))
```

```python
plt.scatter(X,y)
plt.plot(X,predictions,color='red')
```

```python
print(modelo.predict([[8]]))   # [9.5]
```
