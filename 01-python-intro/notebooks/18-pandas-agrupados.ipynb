```python
import pandas as pd

ventas = pd.read_csv(r"C:\Users\SAHUA\Documents\datascience-g8\01-python-intro\data\ventas.csv",
    dtype={"precio_unitario":"float64"})
ventas["import"]=ventas["unidades"]*ventas["precio_unitario"]
```

```python
ventas.info()
```

```python
ventas
```

```python
ventas.groupby("region")["import"].sum()
```

```python
ventas[ventas["region"]=="Sur"]["import"].sum()
```

```python
ventas.groupby("region")["import"].agg(["sum","mean","count"])
```

```python
ventas.groupby("region")["import"].agg(suma="sum",promedio="mean",cantidad="count")
```

```python
ventas.groupby(["region","categoria"])["import"].agg(suma="sum",promedio="mean",cantidad="count")
```

```python
ventas.groupby("region").agg({
    "import":["sum","mean","max"],
    "unidades":["sum"],
    "producto":["nunique"]
})
```
