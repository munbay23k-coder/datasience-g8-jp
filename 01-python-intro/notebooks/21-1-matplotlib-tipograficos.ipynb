## Graficos de linea
- Se usa cuando el eje X sigue un patron

```python
from matplotlib import pyplot as plt
```

```python
x=[1,2,3,4,5]
y=[10,20,25,30,45]

plt.plot(x,y,marker='o',linestyle='-',color='b')
plt.xlabel('Periodo')
plt.ylabel('Valor')
plt.title('Ejemplo de grafico de lineas')
plt.show()
```

## Grafico de barras
- Sirve para comprar magnitudes y responder preguntas como que categoria vende mas?, que clase tiene mayor promedio?, etc
- Compara cantidades

```python
categorias=['A','B','C','D']
valores=[40,70,30,90]

plt.bar(categorias,valores,color=['red','black','blue','green'])
plt.xlabel('Categoria')
plt.ylabel('Frecuencia')
plt.title('Grafico de barras verticales')
plt.show()
```

```python
plt.barh(categorias,valores,color="#25D47D")
plt.xlabel('Frecuencia')
plt.ylabel('Categorias')
plt.title('Grafico de barras verticales')
plt.show()
```

## Grafico pastel o pie chart

```python
frutas=['Manzanas','Platanos','Peras']
porcentaje=[30,50,20]

plt.pie(porcentaje, labels=frutas, autopct='%1.2f%%')## .2 es la cantidad de decimales . f es flotantes
plt.title('Grafico de pastel')
plt.show()
```

## Grafico de dispersion (scatter plot)
- Este grafico me permite visualizar la correlacion , fuerza y relacion entre 2 variables y tmabien puede mostrarme valores atipicos

```python
import numpy as np
x=np.random.rand(50)
y=np.random.rand(50)

plt.scatter(x,y, color='blue')
plt.xlabel('Variable (x)')
plt.ylabel('Variable (y)')
plt.title('Grafico de dispersion')
plt.show()
```

## Grafico histograma
- Reconocer concentracion , dispersion y asimetria. El numero de bins es importante, muy pocos bins me quita detalle, pero muchos bins me da ruido visual

```python
datos = np.random.randn(1000)
plt.hist(datos,bins=20, color='orange',edgecolor='black')
plt.xlabel('Valor')
plt.ylabel('Frecuencia')
plt.title('Histograma')
plt.show()
```

## Grafico de caja, bigotes o boxpot

```python
data_box=[np.random.normal(0,std,100) for std in range(1,4)]

plt.boxplot(data_box,orientation='vertical',patch_artist=True)
plt.xlabel('Grupo')
plt.ylabel('Valor')
plt.title('Grafico Boxpot')
plt.show()
```
