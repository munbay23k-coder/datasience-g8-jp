### Lista VS Array

```python
array= np.array([1,2,3,5])
type(array)
```

```python
array *2 
```

```python
lista = [1,2,3,5]
arr_lista=np.array(lista)
print(arr_lista)
print(type(arr_lista))
```

```python
arr_lista.dtype##para saber el tipo de los elementos
```

```python
lista_decimales=[1.0,2.1,2.4,5.6]
arr_lista=np.array(lista_decimales)
arr_lista.dtype
```

```python
lista = [1,2,3,5]
arr_lista=np.array(lista,dtype='float64')
print(arr_lista)
arr_lista.dtype
```

### Convertir tipo de datos de mi array

```python
lista = [1,2,3,5]
arr_lista=np.array(lista)
print(arr_lista)
arr_lista.dtype
```

```python
arr_float= arr_lista.astype('float64')
print(arr_float)
arr_float.dtype
```

```python
arr_str= arr_lista.astype('str')
print(arr_str)
arr_str.dtype
```
