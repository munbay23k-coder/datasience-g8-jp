```python
###ejercicio 2

stock = np.array(
[
    [25, 18, 30, 12, 20],
    [14, 22, 16, 28, 19],
    [35, 17, 24, 15, 21]
])
ventas = np.array(
    [[10, 20, 12, 8, 7],
    [16, 10, 18, 9, 20],
    [15, 19, 8, 17, 11]])
# 1. Calcula el inventario restante mediante una operación entre matrices.
stock_restante=stock-ventas
print(stock_restante)
print('-'*20)
# 2. Localiza los valores negativos, que representan ventas mayores al stock disponible.
mascara=stock_restante<0
print(stock_restante[mascara])
print('-'*20)
# 3. Crea una copia corregida en la que los valores negativos sean reemplazados por 0.
stock_restante_corregido=stock_restante.copy()
stock_restante_corregido[mascara]=0
print(stock_restante_corregido)
print('-'*20)
# 4. Calcula el stock restante total por categoría antes y después de la corrección.
stock_restante_antes=stock_restante.sum(axis=1)
print(stock_restante_antes)

stock_restante_despues=stock_restante_corregido.sum(axis=1)
print(stock_restante_despues)
print('-'*20)
# 5. Determina la categoría con mayor stock disponible después de la corrección.

print(stock_restante_despues.argmax())
```

```python
## ejercicio 3
precios = np.array(
    [[4.5, 6.0, 3.5, 8.0],
    [4.7, 5.8, 3.6, 8.2],
    [4.6, 6.1, 3.4, 7.9]])
unidades = np.array(
    [[30, 18, 42, 15],
    [25, 20, 35, 17],
    [28, 22, 40, 14]])
# 1. Calcula la facturación por producto y sucursal mediante multiplicación elemento a elemento.
total= precios*unidades
print(total)
print('-'*20)
# 2. Calcula la facturación total de cada sucursal y la facturación total de cada producto.
total_sucursal=total.sum(axis=1)
total_producto=total.sum(axis=0)
print(total_sucursal)
print(total_producto)
print('-'*20)
print('-'*20)
# 3. Identifica la sucursal con mayor facturación.
print(total_sucursal.argmax())
print('-'*20)
# 4. Encuentra el valor máximo de facturación individual y su posición dentro de la matriz.
valor_maximo=total.max()
indixe=total.argmax()
fila, columna = np.unravel_index(indixe, total.shape)
print(fila,columna)
print('-'*20)
for fila in range(total.shape[0]):
    for columna in range(total.shape[1]):
        if valor_maximo==total[fila,columna]:
            print(fila,columna)
print('-'*20)
# 5. Obtén la transpuesta de la matriz de facturación e interpreta qué representan ahora sus filas.
print(total.T)
print('-'*20)
```

```python
##### Ejercicio 4 
notas = np.array(
    [
        [14, 16, 13, 15], 
        [18, 17, 16, 19], 
        [11, 13, 12, 14], 
        [16, 15, 17, 16], 
        [13, 10, 15, 12]
    ]
)
print(notas.shape)
# 1.Calcula el puntaje total y el promedio de cada estudiante sin usar np.mean. 
puntaje_total= notas.sum(axis=1)
print(puntaje_total)
promedio=notas.mean(axis=1)
print(promedio)

promedio_sin_mean=puntaje_total/notas.shape[1]
print(promedio_sin_mean)

# 2.Identifica al estudiante con mayor promedio. 
mejor_estudiante=promedio.argmax()
print(mejor_estudiante)
# 3.Calcula el total acumulado de cada evaluación y determina cuál tuvo menor rendimiento global. 
puntaje_total_materia=notas.sum(axis=0)
print(puntaje_total_materia)
menor_puntaje_materia=puntaje_total_materia.argmin()
print(menor_puntaje_materia)

# 4.Filtra todas las notas menores que 13. 
mascara=notas>13
print(mascara)
notas_mayores_13=notas[mascara]
print(notas_mayores_13)
# 5.Crea una copia y reemplaza las notas menores que 13 por 13; luego recalcula los promedios. 
notas_copia=notas.copy()
mascara=notas<13
notas[mascara]=13
promedios_corregidos=notas.mean(axis=1)
print('-'*20)
print(notas_copia)
print('-'*20)
print(notas)
print('-'*20)
print(promedio)
print(promedios_corregidos)
```

```python
#Ejercicio 5
pasajeros = np.array(
    [
        [120, 135, 128, 142, 150, 180, 175], 
        [95, 102, 110, 108, 115, 130, 125], 
        [160, 155, 170, 168, 172, 190, 185], 
        [80, 88, 92, 100, 105, 118, 112]
    ]
)
# 1. Calcula el total semanal de cada ruta. 
total_semanal=pasajeros.sum(axis=1)
print(total_semanal)
print('-'*20)
# 2. Calcula el total de pasajeros por día considerando todas las rutas. 
total_pasajeros_dia=pasajeros.sum(axis=0)
print(total_pasajeros_dia)
print('-'*20)
# 3. Identifica la ruta con mayor total semanal y el día con mayor demanda global. 
mayor_ruta_x_semana=total_semanal.argmax()
dia_mayor_demanda=total_pasajeros_dia.argmax()
print(mayor_ruta_x_semana)
print(dia_mayor_demanda)
print('-'*20)
# 4. Extrae mediante slicing únicamente los datos del fin de semana (dos últimas columnas). 
datos_fin_semana=pasajeros[::,5:]
print(datos_fin_semana)
print('-'*20)
# 5. Ordena una copia de los totales semanales de menor a mayor. 
total_semanal_copia=total_semanal.copy()
total_semanal_copia.sort()
print(total_semanal)
print(total_semanal_copia)
```

```python
#Ejercicio 11
lecturas = [42, 45, 44, 47, 43, 46, 48, 45, 44, 47, 49, 46, 51, 50, 53, 52, 49, 54, 55, 52, 50, 53, 56, 54] 
# 1. Convierte la lista en un tensor con forma (2, 3, 4) y muestra sus dimensiones. 
lecturas_tensor=np.array(lecturas).reshape((2,3,4))
print(lecturas_tensor)
print('-'*20)
# 2. Obtén todas las lecturas de la segunda máquina. 
segunda_maquina_data=lecturas_tensor[1,:,:]
print(segunda_maquina_data)
print('-'*20)
# 3. Calcula el total de lecturas por máquina realizando sumas por ejes. 
total_por_dia=lecturas_tensor.sum(axis=2)
print(total_por_dia)

total_por_maquina=total_por_dia.sum(axis=1)
print(total_por_maquina)
print(lecturas_tensor.sum(axis=(2,1)))
print('-'*20)
# 4. Encuentra el valor máximo global y su posición dentro del tensor convertido temporalmente en un vector. 
vector=lecturas_tensor.reshape(lecturas_tensor.size)
print(vector)
max_global=vector.max()
print(max_global)
posicion=vector.argmax()
print(posicion)
print('-'*20)
# 5. Reorganiza el tensor como una matriz de 6 × 4, donde cada fila represente un día de una máquina. 
matriz_6_4=lecturas_tensor.reshape(6,4)
print(matriz_6_4)
print('-'*20)
```

```python
# Ejercicio 12
agua = np.array(
    [
        [420, 380, 450, 400, 390], 
        [510, 480, 495, 505, 470], 
        [360, 340, 375, 355, 365], 
        [460, 440, 430, 455, 445]
    ]
) 
# 1. Convierte el array a tipo float64. 
agua_float=agua.astype('float64')## =agua.astype(np.float64)
print(agua.dtype)
print(agua_float.dtype)
print(agua_float.shape)
print('-'*20)
print(agua_float)
print('-'*20)
# 2. Calcula la nueva matriz de consumo aplicando una reducción de 12%.
agua_reducida=agua_float*(1-0.12)
print(agua_reducida)
print('-'*20)
# 3. Calcula los litros ahorrados por parcela y el ahorro total. 
agua_ahorrada=(agua_float-agua_reducida).sum(axis=1)
print(agua_ahorrada)
print(agua_ahorrada.sum())
print('-'*20)
# 4. Identifica la parcela que ahorra más agua. 
parcela_mas_ahorradora=agua_ahorrada.argmax()
print(parcela_mas_ahorradora)
print('-'*20)
# 5. Filtra las zonas cuyo consumo original fue mayor que 450 litros. 

print(agua_float>450)
zonas_mayores450=agua_float[agua_float>450]
print(zonas_mayores450)
print('-'*20)
```

```python
#Ejercicio 16
# Contexto: Un laboratorio necesita construir una matriz base de 5 × 5. Todos los valores parten de 10, pero 
# la diagonal principal representa puntos de calibración y debe valer 110. 
# Datos iniciales 
# Construye la matriz únicamente con np.ones, np.eye y operaciones aritméticas. 

# 1. Crea la matriz solicitada sin escribir manualmente sus 25 valores. 
matriz_base=np.ones((5,5))*10
print(matriz_base)
matriz_nueva=np.eye((5))*100
print(matriz_nueva)
matriz_pedida=matriz_base+matriz_nueva
print(matriz_pedida)
print('-'*30)
# 2. Calcula las sumas por fila y por columna. 
suma_fila=matriz_pedida.sum(axis=1)
suma_colum=matriz_pedida.sum(axis=0)
print(suma_fila)
print(suma_colum)
print('-'*30)
# 3. Comprueba mediante una operación que la matriz es igual a su transpuesta. 
mascara=matriz_pedida==matriz_pedida.T
print(mascara.sum())
print('-'*30)
# 4. Obtén sus valores mínimo y máximo. 
print(matriz_pedida.max())
print(matriz_pedida.min())
print('-'*30)
# 5. Convierte la matriz a tipo int y muestra el tipo de dato final.
matriz_int=matriz_pedida.astype('int8')
print(matriz_int)
print(matriz_int.dtype)
print('-'*30)
```
