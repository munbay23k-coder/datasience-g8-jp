```python
import matplotlib

print(matplotlib.__version__)
```

```python
import numpy as np

x=np.linspace(0,5,10)
y=x**2
```

```python
import matplotlib.pyplot as plt
```

```python
plt.plot(x,y)
plt.show()
```

```python
plt.plot(x,y,'b')
plt.show()
```

```python
plt.plot(x,y,'ro') # r es red y o es puntitos
plt.show()
```

```python
plt.plot(x,y,'ro-') # r es red y o es puntitos
plt.show()
```

```python
plt.plot(x,y,'gD') # r es red y o es puntitos
plt.show()
```

```python
plt.plot(x,y,'gD:') # r es red y o es puntitos
plt.show()
```

```python
plt.plot(x,y,color="#F59F27",linestyle='--',marker='o') 
plt.show()
```

```python
plt.figure(figsize=(12,8))#12 cm de alto y 8 cm de ancho
plt.plot(x,y,color="#F59F27",linestyle='--',linewidth=4,marker='o') 
plt.show()
```

```python
plt.plot(x,y,color="#F59F27",linestyle='--',linewidth=4,marker='o',markerfacecolor='b') 
plt.show()
```
