```python
import numpy as np
from matplotlib import pyplot as plt
```

```python
x = np.linspace(0,5,11)
y1=x**2
y2=x**3

frutas=['Manzanas','Platanos','Peras']
porcentaje=[30,50,20]

categorias=['A','B','C','D']
valores=[40,70,30,90]

plt.figure(figsize=(12,8))
plt.subplot(2,2,1)
plt.plot(x,y1,'bo-')
plt.title('Funcion cuadrada')

plt.subplot(2,2,2)
plt.plot(x,y2,'go-')
plt.title('Funcion cubica')

plt.subplot(2,2,3)
plt.pie(porcentaje, labels=frutas, autopct='%1.2f%%')## .2 es la cantidad de decimales . f es flotantes
plt.title('Grafico de pastel')

plt.subplot(2,2,4)
plt.bar(categorias,valores,color=['red','black','blue','green'])
plt.xlabel('Categoria')
plt.ylabel('Frecuencia')
plt.title('Grafico de barras verticales')

plt.tight_layout(pad=0.01,h_pad=5,w_pad=5)
plt.show()
```
