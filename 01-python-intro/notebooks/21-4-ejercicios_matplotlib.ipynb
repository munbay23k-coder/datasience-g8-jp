```python
# 2.Línea:secuencia de observaciones 
# Grafica con una línea la variable `sepal_length_cm` de las primeras 30 observaciones. Usa el índice de 
# la fila como eje X. Agrega título y etiquetas a ambos ejes. Explica en una frase qué representa el orden 
# en este caso.

subconjunto =df.loc[:29,:]
subconjunto.shape
plt.figure(figsize=(8,4))
plt.plot(subconjunto.index,subconjunto['sepal_length_cm'])
plt.title("Longitud del sepalo (cm) - 30 samples")
plt.xlabel("Indice de observacion")
plt.ylabel("Longitud (cm)")
plt.grid(True,linestyle='--',axis='both',color='green',linewidth=1)
plt.show()
```

```python
# 3.Línea:personalización 
# Repite el gráfico anterior usando marcador circular, línea discontinua y un color diferente. Ajusta el 
# tamaño de la figura y agrega una cuadrícula suave.
plt.figure(figsize=(9,4))
plt.plot(subconjunto.index,subconjunto['sepal_length_cm'],marker='o',linestyle='--')
plt.title("Longitud del sepalo (cm) - 30 samples")
plt.xlabel("Indice de observacion")
plt.ylabel("Longitud (cm)")
plt.grid(True,linestyle='--',axis='both',color='green',linewidth=1,alpha=1)
plt.show()
```

```python
# 4.Barras:promedio por especie 
# Calcula el promedio de `sepal_length_cm` para cada `species` y represéntalo con un gráfico de barras 
# verticales. Incluye etiquetas de ejes y título. 

promedio_sepal_lenght = df.groupby("species")["sepal_length_cm"].mean()
plt.bar(promedio_sepal_lenght.index,promedio_sepal_lenght.values)
plt.title("Promedio por especie")
plt.xlabel("Especie")
plt.ylabel("Promedio")
plt.plot()
```

```python
# 5.Barras horizontales 
# Calcula el promedio de `petal_length_cm` por especie y crea un gráfico de barras horizontales. Ordena 
# las especies de menor a mayor promedio antes de graficar.

promedio_petal_lenght = df.groupby("species")["petal_length_cm"].mean()
plt.barh(promedio_petal_lenght.index,promedio_petal_lenght.values)
plt.title("Longitud promedio por especie")
plt.xlabel("Especie")
plt.ylabel("Promedio")
plt.plot()
```

```python
# 6.Pastel:composición del dataset 
# Calcula cuántas observaciones existen por especie y represéntalas con un gráfico de pastel. Muestra los 
# porcentajes con un decimal. ¿Qué conclusión obtienes sobre el balance del dataset? 

counts=df["species"].value_counts()
plt.figure(figsize=(6,6))
plt.pie(counts.values, labels=counts.index, autopct='%1.1f%%')
plt.title("Proporcion de observaciones por especie")
plt.show()
```

```python
# 7.Dispersión: sépalo 
# Crea un scatter plot entre `sepal_length_cm` y `sepal_width_cm`. Agrega título y etiquetas. Describe si 
# visualmente parece existir una relación clara entre ambas variables. 

plt.figure(figsize=(7,5))
plt.scatter(df["sepal_length_cm"],df["sepal_width_cm"], color='blue')
plt.xlabel('Longitud(cm)')
plt.ylabel('Ancho(cm)')
plt.title('Longitud vs ancho del sepalo')
plt.grid(True,linestyle='--',axis='both',color='green',linewidth=1,alpha=1)
plt.show()
```

```python
# 8.Dispersión por especie 
# Crea un gráfico de dispersión entre `petal_length_cm` y `petal_width_cm`, diferenciando cada especie 
# con una serie distinta y una leyenda. No uses Seaborn.

plt.figure(figsize=(7,5))

for species, group in df.groupby("species"):
    plt.scatter(
       group["petal_length_cm"],
       group["petal_width_cm"],
       label=species,

    )

plt.xlabel('Longitud(cm)')
plt.ylabel('Ancho(cm)')
plt.title('Longitud vs ancho del sepalo por especie')
plt.legend(title="Especie")
plt.grid(True,linestyle='--',axis='both',color='green',linewidth=1,alpha=1)
plt.show()
```

```python
# 9. Histograma: longitud del sépalo 
# Crea un histograma de `sepal_length_cm` con 10 bins, borde visible y etiquetas. Describe dónde se 
# concentra la mayor parte de las observaciones. 
plt.hist(df["sepal_length_cm"],bins=10, color='orange',edgecolor='black')
plt.xlabel('length(cm)')
plt.ylabel('Frecuencia')
plt.title('Histograma length(cm)')
plt.show()
```

```python
# 10.Histograma y bins 
# Crea dos histogramas de `petal_length_cm`: uno con 5 bins y otro con 20 bins. Colócalos lado a lado 
# usando subplots. Explica brevemente cómo cambia la interpretación al modificar los bins.
plt.figure(figsize=(12,8))
plt.subplot(1,2,1)
plt.hist(df["petal_length_cm"],bins=5, color='orange',edgecolor='black')
plt.xlabel('length(cm)')
plt.ylabel('Frecuencia')
plt.title('Histograma petal length(cm)')

plt.subplot(1,2,2)
plt.hist(df["petal_length_cm"],bins=20, color='orange',edgecolor='black')
plt.xlabel('length(cm)')
plt.ylabel('Frecuencia')
plt.title('Histograma petal length(cm)')
plt.show()

## conclusiones: En ciertos valores si no se tuviera el bin 20 se consideraria que hay de varias medidas con valores pero no es el caso y muchas veces solo un valor es representativo o con mayor frecuencia en ese sector
```

```python
# 11.Boxplot por especie 
# Construye un boxplot de `sepal_length_cm` separado por especie. Etiqueta cada caja con el nombre de 
# su especie y compara visualmente las medianas.

plt.figure(figsize=(7,5))

grupos = [
    group["sepal_length_cm"]
    for species, group in df.groupby("species")
]

especies = df["species"].unique()

plt.boxplot(
    grupos,
    tick_labels=especies
)
#plt.boxplot(df["sepal_length_cm"],orientation='vertical',patch_artist=True)
plt.xlabel('Grupo')
plt.ylabel('Valor')
plt.title('Grafico Boxpot')
plt.show()
```

```python
# 12.Boxplot de variables numéricas 
# Crea un solo boxplot que compare las cuatro variables numéricas del dataset. Usa etiquetas cortas y 
# legibles para cada caja. ¿Qué problema de escala observas? 

grupos = [df[col] for col in df.select_dtypes("number").columns]
variable = df.select_dtypes("number").columns
plt.boxplot(grupos,tick_labels=variable)
plt.xlabel('Grupo')
plt.ylabel('Valor')
plt.title('Grafico Boxpot')
plt.show()
##
## El problema que veria seria que los valores de sepal length es ta muy lejos de petal width y esto hace que para saber con mayor precision los valores estos no sean apreciables
```
