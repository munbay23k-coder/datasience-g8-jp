```python
from typing import Any
def extract_text(ruta:str)->dict[str]:
    mi_lista = []
    with open(ruta,"r") as f:
        contenido=f.read()
        lineas = contenido.split("\n")
    for linea in lineas:
        if "registro" in linea.lower():
            continue
        if "id" in linea.lower():
            continue
        if not linea:
            continue
        if "instrucciones" in linea.lower():
            break
        if "lee" in linea.lower():
            break

        mi_lista.append(linea)

    return mi_lista

contenido=extract_text(
    #
    "../data/ventas_polos_aqp.txt"## ".." es para indicar que tome desde donde esta el archivo
)
```

```python
class Pedido:
    def __init__(self,id:str,producto:str,talla:str,color:str,cliente:str,ciudad:str,cantidad:int,precio_unitario:float,metodo_pago:str,estado:str)->None:
        self.id=id
        self.producto=producto
        self.talla=talla
        self.color=color
        self.cliente=cliente
        self.ciudad=ciudad
        self.cantidad=cantidad
        self.precio_unitario=precio_unitario
        self.metodo_pago=metodo_pago
        self.estado=estado
class Tienda:
    def __init__(self,pedidos:dict[str])->None:
        self.pedidos: list[Pedido]=[]

        for pedido in pedidos:
            info = [item.strip() for item in pedido.split(",")]
            pedido = Pedido(
                id= info[0],
                producto = info[1],
                talla =info[2],
                color = info[3],
                cliente= info[4],
                ciudad=info[5],
                cantidad= int(info[6]),
                precio_unitario= float(info[7]),
                metodo_pago= info[8],
                estado=info[9]

            )
            self.pedidos.append(pedido)    

    def calcular_venta(self)->None:
        for pedido in self.pedidos:
            print(f"el pedido con ID:{pedido.id} su monto total es:{pedido.cantidad*pedido.precio_unitario}")

#Muestra cuántas ventas fueron entregadas, pendientes y canceladas.
    def estado_ventas(self)->None:
        entregado=0
        pendiente=0
        cancelado=0
        for pedido in self.pedidos:
            if pedido.estado.lower() == "entregado":
                entregado+=1
            if pedido.estado.lower() == "cancelado":
                cancelado+=1
            if pedido.estado.lower() == "pendiente":
                pendiente+=1
        print(f"Las ventas que fueron entregadas son:{entregado} ")
        print(f"Las ventas que fueron canceladas son:{cancelado} ")
        print(f"Las ventas que fueron pendiente son:{pendiente} ")
#Calcula el ingreso total solo de las ventas entregadas.
    def ventas_entregadas(self)->None:
        ingreso_total=0
        for pedido in self.pedidos:
            if pedido.estado.lower() == "entregado":
                ingreso_total=ingreso_total + int(pedido.cantidad) + int(pedido.precio_unitario) 
        print(f"El monto de ventas entregadas es: {ingreso_total}")
#Muestra cuál fue el producto más vendido por cantidad.
    def producto_mas_vendido(self)->None:
        cantidad_mayor=0
        for pedido in self.pedidos:
            if cantidad_mayor < pedido.cantidad:
                cantidad_mayor=pedido.cantidad
                num_id=pedido.producto
        print(f"El pedido mas vendido es:{num_id}")
#Muestra cuántas ventas se hicieron en Arequipa.
    def ventas_en_arequipa(self)->None:
        cantidad_de_ventas=0
        for pedido in self.pedidos:
            if pedido.ciudad.lower() == "arequipa":
                cantidad_de_ventas+=1
        print(f"La cantidad de ventas en Arequipa es:{cantidad_de_ventas}")
#Muestra el método de pago más usado.
    def metodo_pago_mas_usado(self)->None:
        yape=0
        plin=0
        efectivo=0
        tarjeta=0
        for pedido in self.pedidos:
            if pedido.metodo_pago.lower() == "yape":
                yape+=1
            if pedido.metodo_pago.lower() == "plin":
                plin+=1
            if pedido.metodo_pago.lower() == "efectivo":
                efectivo+=1
            if pedido.metodo_pago.lower() == "tarjeta":
                tarjeta+=1
        dicc_valores={
            "yape":yape,
            "plin":plin,
            "efectivo":efectivo,
            "tarjeta":tarjeta
        }
        ordenado=dict(sorted(dicc_valores.items(),key=lambda x: x[1] , reverse=True))
        print(f"El metodo de pago mas usado es:{list(ordenado.items())[0][0]} ")
#Reto extra: crea un resumen por producto con unidades vendidas e ingreso generado.
    def resumen_pedidos(self)->None:
        for pedido in self.pedidos:
            ingreso_generado=pedido.cantidad*pedido.precio_unitario
            print(f"El {pedido.producto} vendio {pedido.cantidad} y genero: {ingreso_generado} soles")

pedido=Tienda(contenido)       
```
