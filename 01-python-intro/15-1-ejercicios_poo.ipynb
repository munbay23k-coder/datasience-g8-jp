```python
# Ejercicio 2
class Rectangulo:
    def __init__(self,base:float,altura:float):
        self.base=base
        self.altura=altura
    def area(self)->float:
        return self.base*self.altura
    
    def perimetro(self)->float:
        return 2*(self.base +self.altura)
    
    def __str__(self)->str:
        return (f"Mi rectangulo de base {self.base} y altura {self.altura}")
        
    
mi_rectangulo=Rectangulo(3,5)
print(mi_rectangulo.area())
print(mi_rectangulo.perimetro())
print(mi_rectangulo)### busca el metodo __str__
```

```python
# Ejercicio 8

class Reproductor:
    def __init__(self)->None:
        self.playlist=[]
        self.posicion = 0

    def agregar_cancion(self,cancion:str)->None:
        self.playlist.append(cancion)

    def cancion_actual(self)-> str:
        if not self.playlist:
            return "No hay canciones en la playlist"
        
        return self.playlist[self.posicion]
    
    def siguiente(self)->str:
        self.posicion+=1
        if self.posicion < len(self.playlist):
            cancion_actual=self.cancion_actual()
        else:
            self.posicion-=1
            cancion_actual=self.cancion_actual()
        return cancion_actual
    
    def anterior(self)->str:
        self.posicion-=1
        if self.posicion >=0:
            cancion_actual=self.cancion_actual()
        else:
            self.posicion+=1
            cancion_actual=self.cancion_actual()
        return cancion_actual
    
mi_reproductor= Reproductor()

mi_reproductor.cancion_actual()
mi_reproductor.siguiente()
mi_reproductor.anterior()
mi_reproductor.agregar_cancion("we are the world")
mi_reproductor.cancion_actual()
        
        
```

```python
# Ejercicio 9

class Motor:
    def __init__(self,tipo:str,caballos:float)->None:
        self.tipo=tipo
        self.caballos=caballos
        self.encendido=False
    
    def arrancar(self)->None:
        if self.encendido :
            return "El motor ya estaba encendido"
        
        self.encendido=True
        return f"El motor {self.tipo} de {self.caballos} HP se encendio"
    
class Vehiculo:
    def __init__(self,marca:str,motor:Motor)->None:
        self.marca=marca
        self.motor= motor
    
    def arrancar(self)->str:
        return self.motor.arrancar()
    
motor=Motor("gasolina",150)
vehiculo=Vehiculo("ford", motor)

print(vehiculo.arrancar())
print(vehiculo.arrancar())

        
```

```python
# Ejercicio 10

class Libro:
    def __init__(self,titulo:str,autor:str)->None:
        self.titulo=titulo
        self.autor=autor
        self.disponible=True
        
class Usuario:
    def __init__(self,nombre:str)->None:
        self.nombre=nombre
        self.libros_prestados=[]

class Biblioteca:
    def __init__(self)->None:
        self.libros=[]

    def agregar_libro(self,libro:Libro)->None:
        self.libros.append(libro)

    def buscar_libro(self,titulo:str)->Libro|None:
        for libro in self.libros:
            if libro.titulo.lower()==titulo.lower():
                return libro
            
    def prestar_libro(self,titulo:str,usuario:Usuario)->str:
        libro=self.buscar_libro(titulo)

        if libro == None:
            return "El libro no se encuentra en la biblioteca"
        if not libro.disponible:
            return "El libro ya fue prestado"
        
        libro.disponible=False
        usuario.libros_prestados.append(libro)

        return f"Libro prestados: {libro.titulo} al usuario: {usuario.nombre}"
    
    def devolver(self,titulo:str,usuario:Usuario)->str:
        libro=self.buscar_libro(titulo)

        if libro == None:
            return "El libro no se encuentra en la biblioteca"
        
        if libro not in usuario.libros_prestados:
            return f"{usuario.nombre} no tiene prestado el libro: {titulo}"
        
        libro.disponible=True
        usuario.libros_prestados.remove(libro)

        return f"Libro : {titulo} fue devuelto por el usuario: {usuario.nombre}"
    
    def catalogo(self)->None:
        for libro in self.libros:
            print(f"titulo:{libro.titulo} | Autor:{libro.autor}")

biblioteca= Biblioteca()
libro_1=Libro("Mi planta de narnaja lima","Jose mauro de vasconcelos")
libro_2=Libro("Quien mato a palomino","Mario vargas") 

biblioteca.agregar_libro(libro_1)
biblioteca.agregar_libro(libro_2)

biblioteca.catalogo()

usuario_1=Usuario("Pedro")
usuario_2=Usuario("Alejandra")

print("="*20)

print(
    biblioteca.prestar_libro(
        "Quien mato a palomino",
        usuario_1
    )
)
print("="*20)
biblioteca.catalogo()
print("="*20)
for libro in usuario_1.libros_prestados:
    print(libro.titulo)
print("="*20)
biblioteca.devolver("Quien mato a palomino",usuario_2)
print("="*20)
biblioteca.devolver("Quien mato a palomino",usuario_1)
print("="*20)
for libro in usuario_1.libros_prestados:
    print(libro.titulo)
```
