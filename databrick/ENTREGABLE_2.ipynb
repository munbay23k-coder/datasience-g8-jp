```sql
USE tienda;
```

```sql
--EJERCICIO 1
SELECT 
    nombre,
    precio
FROM productos
ORDER BY precio DESC
```

```sql
--EJERCICIO 2
SELECT nombre
FROM clientes
WHERE nombre LIKE 'M%'
```

```sql
--EJERCICIO 3
SELECT *
FROM productos
WHERE stock > 0 AND precio IN(20,60)
```

```sql
--EJERCICIO 4
SELECT *, 
    COALESCE(telefono,'sin dato') as nuevo_telefono
FROM clientes
```

```sql
--EJERCICIO 5
SELECT nombre,precio,
    precio*1.21 AS precio_con_iva
FROM productos
ORDER BY precio DESC
LIMIT 10
```

```sql
--EJERCICIO 6
SELECT 
    nombre,
    precio,
    CASE 
        WHEN precio > 10 THEN 'caro'
        WHEN precio > 5 THEN 'medio'
        ELSE 'barato'
    END AS clasificacion

FROM productos;
```

```sql
--EJERCICIO 7
SELECT DISTINCT pais
FROM clientes;
```
